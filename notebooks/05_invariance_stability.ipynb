# Invariance and stability testing

The maintained assumption is a regime-invariant impact matrix $B$: regimes shift shock variances, not the transmission mechanism. Three checks (spec Section 6).

```python
import numpy as np
import msid
from msid.datasets.simulate import default_msvecm_params, simulate_msvecm

pars = default_msvecm_params(M=3)
y, _ = simulate_msvecm(T=900, alpha=pars['alpha'], beta=pars['beta'],
                       Gammas=pars['Gammas'], B=pars['B'],
                       Lambdas=pars['Lambdas'], P=pars['P'], random_state=7)
res3 = msid.MSVECM(y, lags=1, coint_rank=1, n_regimes=3).fit(
    n_starts=10, random_state=0)
```

## 1. LR test for state-invariant B (M >= 3)

```python
print(res3.test_b_invariance())
```

## 2. Bootstrap overidentification J-test (Tether App. F.1)

```python
print(res3.test_overidentification(n_boot=500, random_state=1))
```

## 3. Overlapping-window Wald test (Tether App. F.2) and rolling coefficient diagnostics (App. E)

```python
res2 = msid.MSVECM(msid.load_example(), lags=1, coint_rank=1,
                   n_regimes=2).fit(n_starts=10, random_state=0)
print(res2.test_b_stability('2020-05-15', window_before=(8, 3),
                            window_after=(3, 8), n_boot=200,
                            min_regime_obs=10, random_state=2))
```

```python
roll = res2.rolling_stability(window=300, lags=2)
roll.plot(equation=res2.model.y.columns[0]);
```
