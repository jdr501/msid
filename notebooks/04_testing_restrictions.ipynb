# Testing competing restriction sets

With MS-identification any zero restriction on $B$ or $\Xi$ is overidentifying and testable. Sign restrictions are normalizations (0 df).

```python
import numpy as np
import msid

y = msid.load_example()
model = msid.MSVECM(y, lags=1, coint_rank=1, n_regimes=2)
res_u = model.fit(n_starts=20, random_state=0)
```

## Identification diagnostics first
LR tests are only as good as the distinctness of the $\lambda$'s.

```python
print(res_u.identification_)
```

## Two competing hypotheses
(Tether-paper style: a 'triangular' set with one sign + one zero, and a 'conventional' set with two zeros.)

```python
R_tri = msid.Restrictions(3).b_zeros([(2, 1)]).b_signs({(2, 0): '+'})
R_conv = msid.Restrictions(3).b_zeros([(2, 0), (2, 1)])
t1, res_tri = res_u.test_restrictions(R_tri, n_starts=10, random_state=1)
t2, res_conv = res_u.test_restrictions(R_conv, n_starts=10, random_state=1)
import pandas as pd
pd.concat([t1, t2], ignore_index=True)
```

## Model comparison table (HL Table 1 layout)

```python
msid.model_table([res_u, res_tri, res_conv],
                 names=['unrestricted', 'triangular', 'conventional'])
```
