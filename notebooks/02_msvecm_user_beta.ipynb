# MSVECM with user-supplied $\beta$

Theory-imposed cointegration vectors skip Johansen estimation entirely -- e.g. HL (2014) impose $\beta = (0,0,0,1)'$ for a stationary interest rate. Here we impose the (approximate) true vector of the example DGP.

```python
import numpy as np
import msid

y = msid.load_example()
beta = np.array([[1.0], [-0.5], [-0.2]])
model = msid.MSVECM(y, lags=1, beta=beta, n_regimes=2)
print(model.beta_source)   # 'user'
res = model.fit(n_starts=20, random_state=0)
res.summary();
```

Compare with the Johansen-based fit: the log-likelihoods should be very close when the imposed vector is (nearly) correct.

```python
res_j = msid.MSVECM(y, lags=1, coint_rank=1, n_regimes=2).fit(
    n_starts=20, random_state=0)
print(f'user beta: {res.loglik_:.2f}   Johansen beta: {res_j.loglik_:.2f}')
```
