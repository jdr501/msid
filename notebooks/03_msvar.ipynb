# MSVAR in levels

The levels VAR is a first-class specification: same EM loop, likelihood, bootstrap and tests as the VECM, with $\Xi = (I - \sum A_i)^{-1} B$ as the long-run matrix (stationarity required).

```python
import numpy as np
import msid
from msid.datasets.simulate import simulate_msvar

A = [np.array([[0.5, 0.1], [0.0, 0.4]])]
B = np.array([[1.0, 0.0], [0.4, 0.8]])
Lambdas = [np.ones(2), np.array([8.0, 0.2])]
P = np.array([[0.95, 0.1], [0.05, 0.9]])
y, states = simulate_msvar(700, A, B, Lambdas, P, random_state=11)
model = msid.MSVAR(y, lags=1, n_regimes=2, deterministic='c')
res = model.fit(n_starts=20, random_state=0)
res.summary();
```

```python
print('long-run matrix Xi:')
print(np.round(res.Xi_, 3))
res.irf(horizon=15, n_boot=200, random_state=2).plot();
```
