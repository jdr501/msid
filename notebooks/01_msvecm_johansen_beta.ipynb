# MSVECM with Johansen $\beta$

Fits a 2-regime Markov-switching SVECM on the bundled example data, with the cointegration vector estimated by Johansen's procedure and held fixed through the EM iterations (the two-step approach of Rajapaksa & Shao 2026).

```python
import numpy as np
import msid

y = msid.load_example()
y.plot(subplots=True, figsize=(9, 6));
```

## Pre-tests
Johansen cointegration test and residual heteroskedasticity diagnostics that motivate the MS specification.

```python
print(msid.johansen(y, det_order=1, k_ar_diff=1))
msid.justify_ms(y, lags=1);
```

## Fit
`coint_rank=None` would select the rank by trace test; here we fix r=1.

```python
model = msid.MSVECM(y, lags=1, coint_rank=1, n_regimes=2, deterministic='ct')
res = model.fit(n_starts=20, random_state=0)
res.summary();
```

## Regimes and impulse responses

```python
res.plot_regimes(series=y.columns[0], threshold=0.7);
```

```python
irf = res.irf(horizon=30, ci=0.68, n_boot=200, random_state=1)
irf.plot();
```

```python
res.fevd(horizon=12, by_state=True).query('horizon == 12')
```
